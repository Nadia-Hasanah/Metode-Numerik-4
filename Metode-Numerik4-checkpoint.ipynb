```python
import math

# Fungsi persamaan nonlinier
def f(t):
    return 800 * math.exp(0.3 * t) - 2000 - 400 * t

# Menentukan batas awal dan toleransi
a = 5
b = 6
toleransi = 0.0001

# Memeriksa interval awal
if f(a) * f(b) >= 0:
    print("Interval tidak memenuhi syarat metode biseksi.")
else:
    iterasi = 0

    # Menampilkan judul tabel
    print("=== METODE BISEKSI ===")
    print(f"{'Iterasi':<10}{'a':<12}{'b':<12}"
          f"{'Tengah':<12}{'f(Tengah)':<15}")

    # Melakukan perhitungan biseksi
    while abs(b - a) > toleransi:
        tengah = (a + b) / 2
        ft = f(tengah)
        iterasi += 1

        # Menampilkan hasil setiap iterasi
        print(f"{iterasi:<10}{a:<12.4f}{b:<12.4f}"
              f"{tengah:<12.4f}{ft:<15.4f}")

        # Memperbarui batas interval
        if f(a) * ft < 0:
            b = tengah
        else:
            a = tengah

    # Menghitung akar persamaan
    akar = (a + b) / 2

    # Menampilkan hasil akhir
    print("\n=== HASIL AKHIR ===")
    print(f"Akar persamaan : {akar:.4f} bulan")
    print(f"Nilai f(akar)  : {f(akar):.4f}")
    print(f"Jumlah iterasi : {iterasi}")
```
